```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import yfinance as yf

# ==========================================
# ⚙️ BACKTEST CONFIGURATION
# ==========================================
TICKERS = ["SOL-USD", "BTC-USD", "ETH-USD"]
TIMEFRAME = "1h"
PERIOD = "730d"  # 2 Years of 1H Data
STARTING_BALANCE = 10000.0  # $10,000 Starting Capital
RISK_PER_TRADE = 0.045  # 4.5% Risk per trade
RR_RATIO = 4.0  # 1:4.0 Risk to Reward Ratio
SL_ATR_MULT = 1.5  # 1.5x ATR Stop Loss


def run_trend_engine_backtest():
  print("🔄 Fetching Historical Data & Running Backtest...")

  overall_balance = STARTING_BALANCE
  equity_curve = [overall_balance]
  trade_history = []

  for ticker in TICKERS:
    print(f"📊 Analyzing {ticker}...")
    df = yf.download(ticker, period=PERIOD, interval=TIMEFRAME, progress=False)

    if df.empty or len(df) < 100:
      continue

    if isinstance(df.columns, pd.MultiIndex):
      df.columns = df.columns.get_level_values(0)

    # Technical Indicators
    df["EMA_50"] = df["Close"].ewm(span=50, adjust=False).mean()
    df["Vol_MA"] = df["Volume"].rolling(20).mean()

    # ATR 14 Calculation
    high_low = df["High"] - df["Low"]
    high_close = np.abs(df["High"] - df["Close"].shift())
    low_close = np.abs(df["Low"] - df["Close"].shift())
    tr = pd.concat([high_low, high_close, low_close], axis=1).max(axis=1)
    df["ATR"] = tr.rolling(14).mean()

    # 20-Period High / Low
    df["High_20"] = df["High"].shift(1).rolling(20).max()
    df["Low_20"] = df["Low"].shift(1).rolling(20).min()

    df = df.dropna()

    # Trade Simulation
    in_position = False
    pos_type = None  # 'LONG' or 'SHORT'
    entry_price = 0.0
    stop_loss = 0.0
    take_profit = 0.0
    position_size = 0.0

    for i in range(2, len(df)):
      current_candle = df.iloc[i]
      prev_candle = df.iloc[i - 1]
      pre_prev_candle = df.iloc[i - 2]

      # Check Exit conditions if in trade
      if in_position:
        if pos_type == "LONG":
          if current_candle["High"] >= take_profit:
            profit = position_size * (take_profit - entry_price)
            overall_balance += profit
            trade_history.append({"ticker": ticker, "pnl": profit, "win": True})
            in_position = False
          elif current_candle["Low"] <= stop_loss:
            loss = position_size * (entry_price - stop_loss)
            overall_balance -= loss
            trade_history.append(
                {"ticker": ticker, "pnl": -loss, "win": False}
            )
            in_position = False

        elif pos_type == "SHORT":
          if current_candle["Low"] <= take_profit:
            profit = position_size * (entry_price - take_profit)
            overall_balance += profit
            trade_history.append({"ticker": ticker, "pnl": profit, "win": True})
            in_position = False
          elif current_candle["High"] >= stop_loss:
            loss = position_size * (stop_loss - entry_price)
            overall_balance -= loss
            trade_history.append(
                {"ticker": ticker, "pnl": -loss, "win": False}
            )
            in_position = False

        equity_curve.append(overall_balance)

      # Check Entry Conditions (Donchian + EMA50 + Volume MA)
      if not in_position:
        bull_cross = (
            (pre_prev_candle["Close"] <= pre_prev_candle["High_20"])
            and (prev_candle["Close"] > prev_candle["High_20"])
            and (prev_candle["Close"] > prev_candle["EMA_50"])
            and (prev_candle["Volume"] > prev_candle["Vol_MA"])
        )

        bear_cross = (
            (pre_prev_candle["Close"] >= pre_prev_candle["Low_20"])
            and (prev_candle["Close"] < prev_candle["Low_20"])
            and (prev_candle["Close"] < prev_candle["EMA_50"])
            and (prev_candle["Volume"] > prev_candle["Vol_MA"])
        )

        if bull_cross:
          entry_price = current_candle["Open"]
          stop_loss = entry_price - (SL_ATR_MULT * prev_candle["ATR"])
          risk_per_unit = entry_price - stop_loss

          if risk_per_unit > 0:
            dollar_risk = overall_balance * RISK_PER_TRADE
            position_size = dollar_risk / risk_per_unit
            take_profit = entry_price + (risk_per_unit * RR_RATIO)
            in_position = True
            pos_type = "LONG"

        elif bear_cross:
          entry_price = current_candle["Open"]
          stop_loss = entry_price + (SL_ATR_MULT * prev_candle["ATR"])
          risk_per_unit = stop_loss - entry_price

          if risk_per_unit > 0:
            dollar_risk = overall_balance * RISK_PER_TRADE
            position_size = dollar_risk / risk_per_unit
            take_profit = entry_price - (risk_per_unit * RR_RATIO)
            in_position = True
            pos_type = "SHORT"

  # ==========================================
  # 📈 PERFORMANCE METRICS & CHART
  # ==========================================
  total_trades = len(trade_history)
  wins = [t for t in trade_history if t["win"]]
  win_rate = (len(wins) / total_trades * 100) if total_trades > 0 else 0
  total_return = (
      (overall_balance - STARTING_BALANCE) / STARTING_BALANCE
  ) * 100

  print("\n" + "=" * 45)
  print("🏆 STRATEGY BACKTEST RESULTS")
  print("=" * 45)
  print(f"Total Trades Taken : {total_trades}")
  print(f"Win Rate           : {win_rate:.1f}%")
  print(f"Initial Capital    : ${STARTING_BALANCE:,.2f}")
  print(f"Final Balance      : ${overall_balance:,.2f}")
  print(f"Total Net Return   : +{total_return:.1f}%")
  print("=" * 45)

  # Plot Portfolio Growth
  plt.figure(figsize=(10, 5))
  plt.plot(
      equity_curve,
      color="#00ff88" if total_return > 0 else "#ff3333",
      linewidth=2,
      label="Portfolio Growth",
  )
  plt.title("Trend Engine Portfolio Growth Curve", fontsize=14, fontweight="bold")
  plt.xlabel("Trade Steps")
  plt.ylabel("Portfolio Balance ($)")
  plt.grid(True, alpha=0.3)
  plt.legend()
  plt.show()


# Run the Backtest
run_trend_engine_backtest()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import yfinance as yf

# ==========================================
# ⚙️ FIXED 2026 BACKTEST SCRIPT
# ==========================================
TICKERS = ["SOL-USD", "BTC-USD", "ETH-USD"]
TIMEFRAME = "1h"
START_DATE = "2026-01-01"
END_DATE = "2026-08-19"
STARTING_BALANCE = 10000.0
RISK_PER_TRADE = 0.045
RR_RATIO = 4.0
SL_ATR_MULT = 1.5

def run_2026_backtest():
  print(f"🔄 Backtesting from {START_DATE} to {END_DATE}...")

  overall_balance = STARTING_BALANCE
  equity_curve = [overall_balance]
  buy_trades = 0
  sell_trades = 0

  for ticker in TICKERS:
    print(f"📊 Analyzing {ticker}...")
    df = yf.download(ticker, start=START_DATE, end=END_DATE, interval=TIMEFRAME, progress=False)

    if df.empty or len(df) < 50:
      continue

    # FIX: Flatten Multi-Index columns if present
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    # Reset index to avoid alignment errors
    df = df.reset_index(drop=True)

    # Setup indicators
    df["EMA_50"] = df["Close"].ewm(span=50, adjust=False).mean()
    df["Vol_MA"] = df["Volume"].rolling(20).mean()

    high_low = df["High"] - df["Low"]
    high_close = np.abs(df["High"] - df["Close"].shift())
    low_close = np.abs(df["Low"] - df["Close"].shift())
    tr = pd.concat([high_low, high_close, low_close], axis=1).max(axis=1)
    df["ATR"] = tr.rolling(14).mean()

    df["High_20"] = df["High"].shift(1).rolling(20).max()
    df["Low_20"] = df["Low"].shift(1).rolling(20).min()
    df = df.dropna()

    in_position = False
    pos_type = None
    entry_price = 0.0
    stop_loss = 0.0
    take_profit = 0.0
    position_size = 0.0

    for i in range(2, len(df)):
      curr = df.iloc[i]
      prev = df.iloc[i - 1]
      pre = df.iloc[i - 2]

      # Exit check
      if in_position:
        if pos_type == "LONG":
          if curr["High"] >= take_profit:
            overall_balance += (position_size * (take_profit - entry_price))
            in_position = False
          elif curr["Low"] <= stop_loss:
            overall_balance -= (position_size * (entry_price - stop_loss))
            in_position = False
        elif pos_type == "SHORT":
          if curr["Low"] <= take_profit:
            overall_balance += (position_size * (entry_price - take_profit))
            in_position = False
          elif curr["High"] >= stop_loss:
            overall_balance -= (position_size * (stop_loss - entry_price))
            in_position = False
        equity_curve.append(overall_balance)

      # Entry check
      if not in_position:
        # Using .values ensures no Index errors
        bull_cross = ((pre["Close"] <= pre["High_20"]) and (prev["Close"] > prev["High_20"]) and (prev["Close"] > prev["EMA_50"]) and (prev["Volume"] > prev["Vol_MA"]))
        bear_cross = ((pre["Close"] >= pre["Low_20"]) and (prev["Close"] < prev["Low_20"]) and (prev["Close"] < prev["EMA_50"]) and (prev["Volume"] > prev["Vol_MA"]))

        if bull_cross:
          entry_price = curr["Open"]
          stop_loss = entry_price - (SL_ATR_MULT * prev["ATR"])
          risk = entry_price - stop_loss
          if risk > 0:
            position_size = (overall_balance * RISK_PER_TRADE) / risk
            take_profit = entry_price + (risk * RR_RATIO)
            in_position = True
            pos_type = "LONG"
            buy_trades += 1
        elif bear_cross:
          entry_price = curr["Open"]
          stop_loss = entry_price + (SL_ATR_MULT * prev["ATR"])
          risk = stop_loss - entry_price
          if risk > 0:
            position_size = (overall_balance * RISK_PER_TRADE) / risk
            take_profit = entry_price - (risk * RR_RATIO)
            in_position = True
            pos_type = "SHORT"
            sell_trades += 1

  print("\n" + "=" * 45)
  print("🏆 2026 YTD PERFORMANCE (JAN 1 - AUG 19)")
  print("=" * 45)
  print(f"Total Buy Trades   : {buy_trades}")
  print(f"Total Sell Trades  : {sell_trades}")
  print(f"Initial Balance    : ${STARTING_BALANCE:,.2f}")
  print(f"Final Balance      : ${overall_balance:,.2f}")
  print(f"Net Return (YTD)   : +{((overall_balance - STARTING_BALANCE)/STARTING_BALANCE)*100:.1f}%")
  print("=" * 45)

run_2026_backtest()
```
